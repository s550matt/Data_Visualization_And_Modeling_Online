```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("Microsoft_Stock.csv")
df.head()
```

# Dataset Overview

This dataset contains historical stock price data for Microsoft. It includes:

Date – Timestamp of the trading day
Open – Opening price
High – Highest price during the day
Low – Lowest price during the day
Close – Closing price
Volume – Number of shares traded

Source: From Kaggle, https://www.kaggle.com/datasets/vijayvvenkitesh/microsoft-stock-time-series-analysis

```python
# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Sort by date (important for time series)
df = df.sort_values(by='Date')
```

```python
plt.figure()
plt.plot(df['Date'], df['Close'])
plt.title("Microsoft Closing Price Over Time")
plt.xlabel("Date")
plt.ylabel("Closing Price")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Visualization #1 - Closing Price Over Time
The closing price shows a clear upward trend over time, indicating overall growth in Microsoft’s stock value. Temporally, the price fluctuates daily, but the long-term trajectory is increasing, reflecting sustained company performance.

```python
plt.figure()
plt.plot(df['Date'], df['Volume'])
plt.title("Microsoft Trading Volume Over Time")
plt.xlabel("Date")
plt.ylabel("Volume")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Visualizaiton #2 - Trading Volume Over Time
Trading volume varies significantly over time, with occasional spikes that may correspond to major market events or announcements. Unlike price, volume does not show a consistent upward or downward trend, highlighting irregular activity across time.

```python
df['Price Range'] = df['High'] - df['Low']

plt.figure()
plt.plot(df['Date'], df['Price Range'])
plt.title("Daily Price Range (Volatility) Over Time")
plt.xlabel("Date")
plt.ylabel("Price Range")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Visualization #3 - High vs Low Price Spread
The price range reflects daily volatility. Over time, volatility fluctuates, with periods of higher instability followed by calmer periods, indicating changing market conditions rather than a consistent spatial (price-based) pattern.
